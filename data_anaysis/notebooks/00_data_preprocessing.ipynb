```python
# Import required libraries
import pandas as pd
import numpy as np
import os

print("✅ Libraries imported successfully")
```

## 1. Preprocessing chickweight.csv

**Original columns**: weight, Time, Chick, Diet  
**Keep only**: weight, Time  
**Remove**: Chick, Diet

```python
# Load chickweight dataset
print("📂 Loading chickweight.csv...")
df_chickweight = pd.read_csv('../data/chickweight.csv')

print(f"✅ Loaded: {df_chickweight.shape[0]} rows, {df_chickweight.shape[1]} columns")
print(f"\n📋 Original columns: {df_chickweight.columns.tolist()}")
print(f"\n📊 First 5 rows:")
df_chickweight.head()
```

```python
# Keep only weight and Time columns
print("🔧 Preprocessing: Keeping only 'weight' and 'Time' columns...")

df_chickweight_clean = df_chickweight[['weight', 'Time']].copy()

print(f"✅ Preprocessed dataset shape: {df_chickweight_clean.shape}")
print(f"📋 Remaining columns: {df_chickweight_clean.columns.tolist()}")
print(f"\n📊 After preprocessing:")
df_chickweight_clean.head(10)
```

```python
# Check for missing values
print("❓ Checking for missing values...")
missing_values = df_chickweight_clean.isnull().sum()
print(missing_values)

if missing_values.sum() == 0:
    print("\n✅ No missing values found!")
else:
    print(f"\n⚠️ Found {missing_values.sum()} missing values")
```

```python
# Display basic statistics
print("📊 Statistics for preprocessed chickweight data:")
df_chickweight_clean.describe()
```

## 2. Preprocessing poultry_farm_dataset.csv

**Original columns**: 14 columns  
**Keep only**: Feed_Consumption, Weight_Growth, Egg_Production  
**Remove**: All other columns

```python
# Load poultry farm dataset
print("📂 Loading poultry_farm_dataset.csv...")
df_poultry = pd.read_csv('../data/poultry_farm_dataset.csv')

print(f"✅ Loaded: {df_poultry.shape[0]} rows, {df_poultry.shape[1]} columns")
print(f"\n📋 Original columns: {df_poultry.columns.tolist()}")
print(f"\n📊 First 5 rows:")
df_poultry.head()
```

```python
# Keep only Feed_Consumption, Weight_Growth, and Egg_Production columns
print("🔧 Preprocessing: Keeping only 'Feed_Consumption', 'Weight_Growth', and 'Egg_Production' columns...")

df_poultry_clean = df_poultry[['Feed_Consumption', 'Weight_Growth', 'Egg_Production']].copy()

print(f"✅ Preprocessed dataset shape: {df_poultry_clean.shape}")
print(f"📋 Remaining columns: {df_poultry_clean.columns.tolist()}")
print(f"\n📊 After preprocessing:")
df_poultry_clean.head(10)
```

```python
# Check for missing values
print("❓ Checking for missing values...")
missing_values = df_poultry_clean.isnull().sum()
print(missing_values)

if missing_values.sum() == 0:
    print("\n✅ No missing values found!")
else:
    print(f"\n⚠️ Found {missing_values.sum()} missing values")
```

```python
# Display basic statistics
print("📊 Statistics for preprocessed poultry farm data:")
df_poultry_clean.describe()
```

## 3. Analyze Correlations

Check the correlation between features in the poultry dataset to understand relationships.

```python
# Check correlation in poultry dataset
print("🔗 Correlation Matrix for Poultry Farm Dataset:")
print("="*60)

correlation_matrix = df_poultry_clean.corr()
print(correlation_matrix)

print("\n📊 Correlation with Feed_Consumption:")
print(correlation_matrix['Feed_Consumption'].sort_values(ascending=False))
```

## 4. Save Preprocessed Datasets

Save the cleaned datasets to the `data/` folder with `_preprocessed` suffix.

```python
# Save chickweight preprocessed dataset
output_path_1 = '../data/chickweight_preprocessed.csv'
df_chickweight_clean.to_csv(output_path_1, index=False)
print(f"✅ Saved: {output_path_1}")
print(f"   Shape: {df_chickweight_clean.shape}")
print(f"   Columns: {df_chickweight_clean.columns.tolist()}")
```
